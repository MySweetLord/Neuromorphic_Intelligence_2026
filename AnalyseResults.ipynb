```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 2 mins sim.
# 2s event intervall
# 20 ordered events

single_run_time = 120 # s

# _____________________________________________________
#          run, ct, cs, wt
# _____________________________________________________

results =  [[1, 15, 4, 1],
            [2, 15, 4, 1],
            [3, 17, 2, 1],
            [4, 4 ,11, 5], 
            [5, 15, 4, 1],
            [6, 14, 5, 1],
            [7, 19, 1, 0],
            [8, 9, 8,  3],
            [9, 4, 11, 5],
            [10,17, 2, 1],
            [11,19, 1, 0]]

# _____________________________________________________


correct_transitions = []
correct_suppresions = []
wrong_transitions   = []

for i, run in enumerate(results):
    correct_transitions.append(run[1])
    correct_suppresions.append(run[2])
    wrong_transitions.append(run[3])

# _____________________________________________________


tot_correct_transitions = np.sum(correct_transitions)
tot_correct_suppresions = np.sum(correct_suppresions)
tot_wrong_transitions   = np.sum(wrong_transitions)

print("tot correct transitions:", tot_correct_transitions)
print("tot correct supressions:", tot_correct_suppresions)
print("tot wrong transitions:", tot_wrong_transitions)

tot_initiated_transitions = sum(correct_transitions) + sum(correct_suppresions) + sum(wrong_transitions)
tot_run_time = len(results)*single_run_time

print("tot inititated transitions:", tot_initiated_transitions)
print("tot run time:")
print(tot_run_time, "s")
print(tot_run_time/60, "min")

# Assuming all wrong transitions are thr0ugh not having changed states when it should

correct_transition_rate = 1- tot_wrong_transitions/(tot_initiated_transitions - tot_correct_suppresions)    # fraction
correct_transition_rate = correct_transition_rate * 100                                                     # percent 

correct_suppresion_rate = tot_correct_suppresions/tot_correct_suppresions
correct_suppresion_rate = correct_suppresion_rate * 100

print(f"correct transition rate: {correct_transition_rate:.1f} %")
print(f"correct suppression rate: {correct_suppresion_rate:.1f} %")

# _____________________________________________________

mean_correct_transitions = np.mean(correct_transitions)
mean_correct_suppresions = np.mean(correct_suppresions)
mean_wrong_transitions   = np.mean(wrong_transitions)


```
